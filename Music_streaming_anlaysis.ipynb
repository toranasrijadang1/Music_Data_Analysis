```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv("music_streaming_habits_2026.csv")
print("First 5 records:")
print(df.heada
```

```python
print("\nMissing Values:")
print(df.isnull().sum())
print("\nDuplicate Records:")
print(df.duplicated().sum())
```

```python
print("\nDuplicate Records:")
print(df.duplicated().sum())
```

```python
print("\nTop Artists:")
artist_count = df["top_artist"].value_counts()
print(artist_count.head(10))
```

```python
print("\nGenre Distribution:")
genre_count = df["top_genre"].value_counts()
print(genre_count)
```

```python
print("\nMost Popular Artists:")
most_played = df.groupby(
    "top_artist"
)["songs_per_day"].mean().sort_values(
    ascending=False
)
print(most_played.head(10))
```

```python
print("\nListening Trend:")
listening_trend = df.groupby(
    "top_genre"
)["daily_listening_minutes"].mean()
print(listening_trend)
```

```python
top_artists = df["top_artist"].value_counts().head(10)
plt.figure(figsize=(10, 5))
plt.bar(
    top_artists.index,
    top_artists.values
)
plt.title("Top 10 Artists")
plt.xlabel("Artist")
plt.ylabel("Number of Listeners")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 8))
plt.pie(
    genre_count.values,
    labels=genre_count.index,
    autopct="%1.1f%%"
)
plt.title("Genre Distribution")
plt.show()
```

```python
plt.figure(figsize=(10, 5))

plt.plot(
    listening_trend.index,
    listening_trend.values,
    marker="o"
)

plt.title("Average Listening Time by Genre")
plt.xlabel("Genre")
plt.ylabel("Average Daily Listening Minutes")

plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```
